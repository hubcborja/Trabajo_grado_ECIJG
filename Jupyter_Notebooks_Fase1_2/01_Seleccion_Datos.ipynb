### Determinación canasta báscia de productos de importación y exportación representativos


```python
# Extraer canasta representativa 51% de las importaciones
imports21_24_sintotal=imports21_24[imports21_24['cmdCode']!='TOTAL']
imports21_24_sintotal=imports21_24_sintotal[~imports21_24_sintotal['partnerISO'].isin(['S19', 'X2 ', 'W00','_X '])]
imports21_24_partida=imports21_24_sintotal[imports21_24_sintotal['cmdCode'].str.len() == 4].copy()
imports21_24_partida=imports21_24_partida[~imports21_24_partida['cmdCode'].isin(['9999'])]
imports21_24_agr=imports21_24_partida.groupby(['cmdCode','cmdDesc'])['cifvalue'].sum().reset_index()
imports21_24_agr=imports21_24_agr.sort_values(by='cifvalue', ascending=False)
total_valorcif_imp=imports21_24_agr['cifvalue'].sum()
imports21_24_agr['part_cifvalue']=imports21_24_agr['cifvalue']/total_valorcif_imp
imports21_24_agr['part_acum']=imports21_24_agr['part_cifvalue'].cumsum()
cesta_imp=imports21_24_agr[imports21_24_agr['part_acum']<=0.51]
lista_cod_cesta_imp=cesta_imp['cmdCode'].tolist()
print(lista_cod_cesta_imp)

# Extraer canasta representativa 51% de las exportaciones
exports21_24_sintotal=exports21_24[exports21_24['cmdCode']!='TOTAL']
exports21_24_sintotal=exports21_24_sintotal[~exports21_24_sintotal['partnerISO'].isin(['S19', 'X2 ', 'W00','_X '])]
exports21_24_partida=exports21_24_sintotal[exports21_24_sintotal['cmdCode'].str.len() == 4].copy()
exports21_24_sintotal=exports21_24_sintotal[~exports21_24_sintotal['cmdCode'].isin(['9999'])]
exports21_24_agr=exports21_24_partida.groupby(['cmdCode','cmdDesc'])['fobvalue'].sum().reset_index()
exports21_24_agr=exports21_24_agr.sort_values(by='fobvalue', ascending=False)
total_valorfob_exp=exports21_24_agr['fobvalue'].sum()
exports21_24_agr['part_fobvalue']=exports21_24_agr['fobvalue']/total_valorfob_exp
exports21_24_agr['part_acum']=exports21_24_agr['part_fobvalue'].cumsum()
cesta_exp=exports21_24_agr[exports21_24_agr['part_acum']<=0.51]
lista_cod_cesta_exp=cesta_exp['cmdCode'].tolist()
print(lista_cod_cesta_exp)
```

```python
# Extraer canasta representativa 80% de los paises que más colombia importa
imports21_24_sintotal=imports21_24[imports21_24['cmdCode']!='TOTAL']
imports21_24_sintotal=imports21_24_sintotal[~imports21_24_sintotal['partnerISO'].isin(['S19', 'X2 ', 'W00','_X '])]
imports21_24_partida=imports21_24_sintotal[imports21_24_sintotal['cmdCode'].str.len() == 4].copy()
imports21_24_partida=imports21_24_partida[~imports21_24_partida['cmdCode'].isin(['9999'])]
imports21_24_agr=imports21_24_partida.groupby(['partnerCode','partnerISO','partnerDesc'])['cifvalue'].sum().reset_index()
imports21_24_agr=imports21_24_agr.sort_values(by='cifvalue', ascending=False)
total_valorcif_imp=imports21_24_agr['cifvalue'].sum()
imports21_24_agr['part_cifvalue']=imports21_24_agr['cifvalue']/total_valorcif_imp
imports21_24_agr['part_acum']=imports21_24_agr['part_cifvalue'].cumsum()
paises_imp=imports21_24_agr[imports21_24_agr['part_acum']<=0.80]
lista_cod_paises_imp=paises_imp['partnerCode'].tolist()
lista_cod_paises_imp_iso=paises_imp['partnerISO'].tolist()
print(lista_cod_paises_imp_iso)

# Extraer canasta representativa 80% de los paises que más colombia exporta
exports21_24_sintotal=exports21_24[exports21_24['cmdCode']!='TOTAL']
exports21_24_sintotal=exports21_24_sintotal[~exports21_24_sintotal['partnerISO'].isin(['S19', 'X2 ', 'W00','_X '])]
exports21_24_partida=exports21_24_sintotal[exports21_24_sintotal['cmdCode'].str.len() == 4].copy()
exports21_24_partida=exports21_24_partida[~exports21_24_partida['cmdCode'].isin(['9999'])]
exports21_24_agr=exports21_24_partida.groupby(['partnerCode','partnerISO','partnerDesc'])['fobvalue'].sum().reset_index()
exports21_24_agr=exports21_24_agr.sort_values(by='fobvalue', ascending=False)
total_valorfob_exp=exports21_24_agr['fobvalue'].sum()
exports21_24_agr['part_fobvalue']=exports21_24_agr['fobvalue']/total_valorfob_exp
exports21_24_agr['part_acum']=exports21_24_agr['part_fobvalue'].cumsum()
paises_exp=exports21_24_agr[exports21_24_agr['part_acum']<=0.80]
lista_cod_paises_exp=paises_exp['partnerCode'].tolist()
lista_cod_paises_exp_iso=paises_exp['partnerISO'].tolist()
print(lista_cod_paises_exp_iso)
```

```python
# Socios representativos (Se agrega Colombia, socios mayoritarios, paises ASEAN, OCDE y los que tienen acuerdos para evitar doble tributacion)
paises_exp_imp_adicionales=[32, 36, 40, 56, 68, 76, 96, 104, 116, 124, 152, 156, 188, 192, 203, 208, 218, 233, 246, 251, 276, 300, 328, 348, 352, 360, 372,
                            376, 380, 392, 410, 418, 428, 440, 442, 458, 484, 528, 554, 579, 591, 600, 604, 608, 616, 620, 626, 699, 702, 703, 704, 705, 724,
                            740, 752, 757, 764, 784, 792, 826, 842, 858, 862]
socios_comerciales=list(set(paises_exp_imp_adicionales + [170]))
# Años
chunks_anios = ['2004,2005,2006','2007,2008,2009','2010,2011,2012','2013,2014,2015','2016,2017,2018', '2019,2020,2021','2022,2023,2024']
# Canasta representativa
lista_cestas=list(set(lista_cod_cesta_exp + lista_cod_cesta_imp))
cmd_cadena = ','.join(str(c) for c in lista_cestas)
```

```python
print('Paises identificados como socios:', len(socios_comerciales))
print('Número de productos en la canasta:', len(lista_cestas))
```

```python
exports_imports_socios=[]
for socio in socios_comerciales:
    for anio in chunks_anios:
        try:
            temp = comtradeapicall.getFinalData(subscription_key=PrimaryKey, typeCode='C', freqCode='A', clCode='HS', period=anio,
                                    reporterCode=str(socio), cmdCode=cmd_cadena, flowCode='M,X', partnerCode=None,partner2Code=None,
                                    customsCode=None, motCode=None, maxRecords=100000, format_output='JSON',
                                    aggregateBy=None, breakdownMode='classic', countOnly=None, includeDesc=True)
            if temp is None or len(temp) == 0:
                print(f"SIN DATOS: {socio} — {anio}")
                continue
            temp=temp[['reporterCode', 'reporterISO', 'partnerCode', 'partnerISO', 'flowCode', 'refYear', 'cmdCode','primaryValue', 'netWgt']]
            exports_imports_socios.append(temp)
            print('Extraído:', temp['reporterISO'].unique(), anio, 'filas:', len(temp))
            time.sleep(4)
        except Exception as e:
            print(f"ERROR REAL en {socio} / {anio}: {type(e).__name__}: {e}")
```

## Enriquecimiento para corregir y verificar los ISO

```python
# Importacion diccionario desde Comtrade para verificar los ISO3 correctos
datos_json = requests.get("https://comtradeapi.un.org/files/v1/app/reference/Reporters.json").json()
df_paises = pd.DataFrame(datos_json['results'])

columnas_reporters = ['id', 'text', 'reporterCodeIsoAlpha3']
columnas_exist = [col for col in columnas_reporters if col in df_paises.columns]

df_diccionario_paises = df_paises[columnas_exist].copy()

df_diccionario_paises.rename(columns={'id': 'partnerCode',
                                      'text': 'country_name',
                                      'reporterCodeIsoAlpha3': 'iso3'}, inplace=True)

print(df_diccionario_paises.head())
```

```python
# merge para garantizar que el Reporter tenga su ISO3 correcto para reporter y partner
base_red = exports_imports_socios_red_commodities.merge(df_diccionario_paises[['partnerCode', 'iso3']],
                                                        left_on='reporterISO',
                                                        right_on='iso3',
                                                        how='left')
base_red = base_red.merge(df_diccionario_paises[['partnerCode', 'iso3']],
                          left_on='partnerISO',
                          right_on='iso3',
                          how='left').rename(columns={'iso3_y': 'partner_iso_limpio',
                                                      'iso3_x': 'reporter_iso_limpio'})
base_red
```

```python
# Se agrupa por año, origen, destino y expo o impo y suma el valor
flujos_consolidados = base_red.groupby(['refYear', 'reporter_iso_limpio', 'partner_iso_limpio', 'flowCode'])['primaryValue'].sum().reset_index()

# Se filtra para los socios de interés
lista_iso_socios = df_diccionario_paises[df_diccionario_paises['partnerCode'].isin(socios_comerciales)]['iso3'].tolist()
flujos_consolidados = flujos_consolidados[(flujos_consolidados['reporter_iso_limpio'].isin(lista_iso_socios))]
flujos_consolidados.head(1)
```

```python
# Pivotar tabla
red_final = flujos_consolidados.pivot_table(index=['refYear', 'reporter_iso_limpio', 'partner_iso_limpio'],
                                            columns='flowCode',
                                            values='primaryValue',
                                            fill_value=0).reset_index()

# Peso para la red
red_final['peso_comercio_total'] = red_final['M'] + red_final['X']
red_final['balanza_comercial'] = red_final['X'] - red_final['M']
red_final['ln_peso_comercio_total'] = np.log1p(red_final['peso_comercio_total'])

red_final=red_final.loc[red_final['reporter_iso_limpio']!=red_final['partner_iso_limpio']]
```
